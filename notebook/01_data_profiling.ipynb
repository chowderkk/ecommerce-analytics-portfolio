Delivery Date Investigation

The Sales dataset contains 49,719 missing Delivery Date values.

Initial investigation showed that the missing values are not randomly distributed. StoreKey = 0 represents the Online sales channel, and all 13,165 online transactions have a Delivery Date.

A cross-tabulation showed:

- StoreKey = 0 (Online): 13,165 transactions with Delivery Date
- StoreKey > 0 (Physical stores): 49,719 transactions without Delivery Date

Therefore, the missing Delivery Date values appear to be structural rather than erroneous. Delivery dates are relevant to online transactions, while physical-store purchases do not require delivery.

The missing values should therefore not be automatically filled or removed during data cleaning.

```python
delivery_by_channel = pd.crosstab(
    sales["StoreKey"] == 0,
    sales["Delivery Date"].isna(),
    normalize="index"
).round(4) * 100

delivery_by_channel
```

Customer Investigation Findings

Further investigation identified 3,379 customers with no corresponding sales transactions. These customers are distributed across all major markets, with the largest groups located in the United States, Australia, Canada, and the United Kingdom. They will not be classified as inactive customers without additional evidence.

The 10 missing State Code values all belong to customers whose State is listed as Napoli, Italy. No other State Code values are associated with Napoli in the dataset, so the missing values cannot be reliably inferred from the available data. They will therefore be retained as NULL rather than replaced with potentially incorrect values.

CustomerKey referential integrity was also validated: every CustomerKey in the Sales table exists in the Customers table.
